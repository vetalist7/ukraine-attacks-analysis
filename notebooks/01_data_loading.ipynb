```python
# Create attack type classification based on model and is_shahed columns

def classify_attack(row):
    model = str(row['model']).lower()
    
    if 'shahed' in model or row.get('is_shahed', 0) == 1:
        return 'drone'
    elif any(x in model for x in ['iskander', 'kalibr', 'kh-', 'x-', 'kinzhal']):
        return 'missile'
    elif 'uav' in model:
        return 'drone'
    else:
        return 'other'

df['attack_type'] = df.apply(classify_attack, axis=1)
```

```python
# Create Clean Data

df['attack_type'] = df['attack_type'].astype(str)
df['date'] = pd.to_datetime(df['time_start'], errors='coerce').dt.date
```

```python
df.applymap(type).head()
```

```python
daily = (
    df.groupby(['date', 'attack_type'])['launched']
    .sum()
    .reset_index()
)
```

```python
daily_pivot = daily.pivot(index='date', columns='attack_type', values='launched').fillna(0)
```
